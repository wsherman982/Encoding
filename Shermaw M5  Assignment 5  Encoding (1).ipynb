```python
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn.model_selection import cross_val_score

bikeDataSet = pd.read_excel(
    r"C:\Users\wsher\OneDrive\Documents\hour.csv.xlsx"
).values

X = bikeDataSet[1:, 2:-1]
y = bikeDataSet[1:, -1]

estimator = linear_model.LinearRegression()
score = cross_val_score(estimator, X, y).mean()

print(f"Score = {score:.3f}")
```

```python
import numpy as np
from sklearn import linear_model
from sklearn.model_selection import cross_val_score

# Create random features
X_random = np.random.random((X.shape[0], 4))

# Use Linear Regression
estimator = linear_model.LinearRegression()

score = cross_val_score(estimator, X_random, y).mean()

print(f"Random Features Score = {score:.3f}")
```

```python
from sklearn import linear_model
from sklearn.model_selection import cross_val_score

# Return to the original bike-sharing features
X = bikeDataSet[1:, 2:-1]
y = bikeDataSet[1:, -1]

estimator = linear_model.Lasso(alpha=0.1)

score = cross_val_score(estimator, X, y).mean()

print(f"Lasso Score = {score:.3f}")
```

```python
The original Linear Regression and Lasso models both produced a score of 1.000, while using random features reduced the score to -0.279. The random features did not provide useful information for predicting the bike-sharing count.
```

```python
from sklearn import datasets
from sklearn.model_selection import cross_val_score
from sklearn.svm import SVC

# Load the Iris dataset
iris = datasets.load_iris()
X = iris.data
y = iris.target

# Try different alpha values
alphas = [0.001, 0.01, 0.1, 1, 10, 100, 1000]

for alpha in alphas:
    X_scaled = X * alpha
    estimator = SVC(kernel='linear')
    score = cross_val_score(estimator, X_scaled, y).mean()
    print(f"Alpha = {alpha}, Score = {score:.3f}")
```

```python
The highest cross-validation score was 0.980, which occurred when alpha = 1.
```
